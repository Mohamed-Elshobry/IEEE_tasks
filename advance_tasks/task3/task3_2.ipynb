```python
import cv2
import matplotlib.pyplot as plt
```

```python
coins = cv2.imread(r"images\SPOILER_coins_image.jpg")
shapes = cv2.imread(r"images\SPOILER_shapes.jpg")
```

```python
def contour_detection(img, threshold1, threshold2):
    contour_img = img.copy()
    img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    Gussing_Blurring = cv2.GaussianBlur(img, (13,5), 0)

    edges = cv2.Canny(Gussing_Blurring, threshold1, threshold2)


    contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cv2.drawContours(contour_img, contours, -1, (0, 255, 0), 3)

    plt.imshow(contour_img)
    plt.title("image contour")
    plt.axis('off')
    plt.show()
```

```python
contour_detection(coins, 50, 150)
```

```python
contour_detection(shapes, 50, 150)
```
